# Graph functions

```python
def print_matrix(vertices, matrix):
  """
  Prints a graph given by the matrix.
  """
  n = len(matrix)   # the order of a graph
  # take the proper names of vertices
  if (vertices is not None) and (len(vertices) == n):
    vv = vertices
  else:
    vv = range(1, n+1)    # numbers 1..n
  # printing
  for i in range(n):
    print(vv[i], ": ", sep = "", end="")
    for j in range(n):
      if matrix[i][j] == 1:
        print(vv[j], end=" ")
    print()

def print_graph(graph):
  """
  Prints a graph given by the adjacency list (or just a dictionary).
  """
  for v in graph:
    print(v, ": ", sep = "", end="")
    for u in graph[v]:
      print(u, end=" ")
    print()
```

```python
# Making and modyfing the graphs

def add_vertex(graph,vertex):
    if vertex not in graph:
        graph[vertex]=[]

def add_arc(graph,arc):
    u,v=arc
    add_vertex(graph,u)
    add_vertex(graph,v)
    if v not in graph[u]:
        graph[u].append(v)

def add_edge(graph,edge):
    u,v=edge
    if u!=v:
        add_arc(graph,(u,v))
        add_arc(graph,(v,u))
    else:
        raise ValueError('Loops are not allowed')
```

# Random graphs generator with $G(n,p)$

```python
from random import random, seed
import numpy as np
seed(2026)
def random_graph(n,p):
    graph={}
    for i in range(1,n+1):
        add_vertex(graph,i)
        for j in range(1,i):
            if random()<p:
                add_edge(graph,(i,j))
    return graph
```

```python
g = random_graph(5, 1/2)
print(g)
```

```python
def graph_to_matrix(graph):
    vertices = list(graph.keys())
    n = len(vertices)
    matrix = np.zeros((n,n), dtype=int)

    for i in range(n):
        for neighbor in graph[vertices[i]]:
            j = vertices.index(neighbor)
            matrix[i][j] = 1

    return matrix, vertices
```

```python
g = {"a": ["b"], "b": ["a"], "c": []}
mat, vertices = graph_to_matrix(g)
assert vertices == ["a", "b", "c"]
assert np.array_equal(mat, [[0,1,0],
                            [1,0,0],
                            [0,0,0]])
mat
```

```python
def matrix_to_graph(vertices, matrix):
    n = len(matrix)   # the order of a graph
    # take the proper names of vertices
    if (vertices is not None) and (len(vertices) == n):
      vv = vertices
    else:
      vv = range(1, n+1)    # numbers 1..n

    for row in matrix:
        if len(row) != n:
            raise ValueError('Matrix must be square')
    graph = {}
    for v in vv:
        add_vertex(graph, v)
    for i in range(n):
        for j in range(n):
            if matrix[i][j] == 1:
                add_arc(graph, (vv[i], vv[j]))
    return graph
```

```python
g2 = matrix_to_graph(vertices, mat)
assert g2 == g
g2
```

```python
def cycle(n):
    if n < 3:
        raise ValueError('Cycle needs at least 3 vertices')

    graph = {}
    for i in range(1, n+1):
        add_vertex(graph, i)
    for i in range(1, n):
        add_edge(graph, (i, i+1))
    add_edge(graph, (n, 1))

    return graph
```

```python
import random
def add_cycle(graph, start, n):
    if not isinstance(graph, dict):
        graph = matrix_to_graph(list(range(1, len(graph)+1)), graph)

    if start not in graph:
        raise ValueError('Start vertex does not exist')

    if n < 3 or n > len(graph):
        raise ValueError('Invalid cycle length')

    vertices = list(graph.keys())
    vertices.remove(start)

    path = [start] + random.sample(vertices, n-1)

    for i in range(n-1):
        add_edge(graph, (path[i], path[i+1]))

    add_edge(graph, (path[-1], start))

    return graph
```

```python
g = {1: [], 2: [], 3: [], 4: []}
g = add_cycle(g, 1, 3)
g
```

```python
def cycle(n):
    if n < 3:
        raise ValueError('Cycle needs at least 3 vertices')

    graph = {i: [] for i in range(1, n+1)}
    return add_cycle(graph, 1, n)
```
