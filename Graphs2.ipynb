# Graph functions

```python
def print_matrix(vertices, matrix):
  """
  Prints a graph given by the matrix.
  """
  n = len(matrix)   # the order of a graph
  # take the proper names of vertices
  if (vertices is not None) and (len(vertices) == n):
    vv = vertices
  else:
    vv = range(1, n+1)    # numbers 1..n
  # printing
  for i in range(n):
    print(vv[i], ": ", sep = "", end="")
    for j in range(n):
      if matrix[i][j] == 1:
        print(vv[j], end=" ")
    print()

def print_graph(graph):
  """
  Prints a graph given by the adjacency list (or just a dictionary).
  """
  for v in graph:
    print(v, ": ", sep = "", end="")
    for u in graph[v]:
      print(u, end=" ")
    print()
```

# Making and modyfying graphs

```python
def add_vertex(graph, vertex):
  """
  Adds a vertex to a graph.
  """
  if vertex not in graph:
    graph[vertex] = []
```

```python
def add_arc(graph, arc):
  """
  Adds an arc to a graph.
  """
  u, v = arc
  add_vertex(graph, u)
  add_vertex(graph, v)
  if v not in graph[u]:
    graph[u].append(v)
```

```python
def add_edge(graph, edge):
  """
  Adds an edge to a graph (simple and undirected) - loops are not allowed.
  """
  u, v = edge
  add_vertex(graph, u)
  add_vertex(graph, v)
  if u == v:
    raise ValueError("Loops are not allowed.")
  if v not in graph[u]:
    graph[u].append(v)
  if u not in graph[v]:
    graph[v].append(u)
```

# Random graph generator with $ G(n,p) $ model.

```python
from random import random, seed
# for repeatness
seed(2026)
```

```python
# prepare graph G(10, 1/3)
n = 10
p = 1/3
random_graph = {}
for i in range(1, n+1):
  add_vertex(random_graph, i)
  for j in range(1, i):
    if random() < p:
        add_edge(random_graph, (i, j))
```

```python
print_graph(random_graph)
```
