# Graph functions

```python
def print_matrix(vertices, matrix):
    """
    Prints a graph given by the matrix.
    """
    n = len(matrix)  # the order of a graph
    # take the proper names of vertices
    if (vertices is not None) and (len(vertices) == n):
        vv = vertices
    else:
        vv = range(1, n + 1)  # numbers 1..n
    # printing
    for i in range(n):
        print(vv[i], ": ", sep="", end="")
        for j in range(n):
            if matrix[i][j] == 1:
                print(vv[j], end=" ")
        print()


def print_graph(graph):
    """
    Prints a graph given by the adjacency list (or just a dictionary).
    """
    for v in graph:
        print(v, ": ", sep="", end="")
        for u in graph[v]:
            print(u, end=" ")
        print()
```

```python
# Making and modifying the graphs

def add_vertex(graph, vertex):
    if vertex not in graph:
        graph[vertex] = []


def add_arc(graph, arc: tuple):
    u, v = arc
    add_vertex(graph, u)
    add_vertex(graph, v)
    if v not in graph[u]:
        graph[u].append(v)


def add_edge(graph, edge):
    u, v = edge
    if u != v:
        add_arc(graph, (u, v))
        add_arc(graph, (v, u))
    else:
        raise ValueError('Loops are not allowed')
```

# Reading and writing from/to a file.

```python
def graph_from_edges(filename, directed=0):
    """
    Reads a graph from a file in which each line contains two (or more) words -
    - the description of an arc (or edge) or just one word (the vertex name).
    Variable filename contains full file path.
    """
    graph = {}
    with open(filename, "r") as f:  # open for reading
        for line in f:
            words = line.strip().split()
            if len(words) == 1:  # the vertex name
                add_vertex(graph, words[0])
            elif len(words) >= 2:  # the arc/edge descriptions - use just first two
                if directed:
                    add_arc(graph, (words[0], words[1]))
                else:
                    add_edge(graph, (words[0], words[1]))
    return graph
```

```python
def graph_to_neighbourlist(graph, filename):
    """
    Writes a graph given by the adjacency list (or just a dictionary) to the textfile.
    """
    with open(filename, "w") as f:  # open for writing
        for v in graph:
            line = f"{v}:"
            for u in graph[v]:
                line += f" {u}"
            line += "\n"
            f.write(line)
```

```python
def graph_from_neighbourlist(filename):
    """
    Returns a graph given by the adjacency list from the text file.
    """
    graph = {}
    with open(filename, 'r') as f:
        for line in f:
            vertex = line[0]  # we take the vertex name
            words = line[2:].strip().split()  # we take the neighbours names

            for v in words:
                add_arc(graph, (vertex, v))
    return graph
```

```python
def graph_to_edges(graph, filename):
    """
    Writes a graph given by the adjacency list (or just a dictionary) to the textfile.
    """
    with open(filename, "w") as f:  # open for writing
        for v in graph:
            if len(graph[v]) == 0:
                line = f'{v}\n'
                f.write(line)
                continue
            for u in graph[v]:
                line = f"{v} {u}\n"
                f.write(line)
```
